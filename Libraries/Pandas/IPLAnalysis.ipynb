```python
# isin()
# Find the most destructive death over batsman in the history of IPL
# Strike Rate= (Number of runs/Number of balls)/100e
# Min batsman 200 balls in over 16-20
```

```python
death_overs=delivery[delivery['over']>15]
death_overs
```

```python
all_batsman=death_overs.groupby('batsman')['batsman_runs'].count()
x=all_batsman>200
batsman_list=all_batsman[x].index.tolist()
```

```python
final=delivery[delivery['batsman'].isin(batsman_list)]
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls=final.groupby('batsman')['batsman_runs'].count()
```

```python
sr=(runs/balls)*100
```

```python
sr.sort_values(ascending=False)
```

```python

#Merge Function

delivery=pd.read_csv('deliveries.csv')
match=pd.read_csv('matches.csv')
```

```python
delivery.head(2)
```

```python
match.head(2)
```

```python
new =delivery.merge(match,left_on='match_id',right_on='id')
```

```python
new.shape
```

```python
season_batsman=new.groupby(['season','batsman'])['batsman_runs'].sum()
```

```python
season_batsman
```

```python
season_batsman.sort_values(ascending=False).reset_index().drop_duplicates(subset='season',keep='first').sort_values('season')
```
