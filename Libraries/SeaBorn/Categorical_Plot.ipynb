# Categorical Plots in Seaborn

## Categorical Scatter Plots
- **Stripplot**: A scatter plot that represents data points along a categorical axis.
- **Swarmplot**: Similar to stripplot but adjusts points to avoid overlap.

## Categorical Distribution Plots
- **Boxplot**: Displays the distribution of data using quartiles and outliers.
- **Violinplot**: Combines aspects of a boxplot and a KDE plot, showing distribution shape.

## Categorical Estimate Plots (for Central Tendency)
- **Barplot**: Shows the mean (or other estimations) of the data with error bars.
- **Pointplot**: Displays point estimates and confidence intervals.
- **Countplot**: Visualizes the count of observations in each categorical bin.

## Figure-Level Function
- **Catplot**: A high-level interface for creating categorical plots.

```python
# import datasets
import seaborn as sns
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

```python
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
sns.catplot(data=tips,kind='scatter', x='total_bill',y='tip')
```

```python
# strip plot
# axes level function
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
# using catplot
# figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
# jitter
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=0.1,hue='sex')
```

```python
# swarmplot
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
sns.swarmplot(data=tips, x='day',y='total_bill')
```

```python
# hue
sns.swarmplot(data=tips, x='day',y='total_bill',hue='sex')
```

Categorical Estimate Plot -> for central tendency

BOXPLOT
A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.



```python
# Box plot
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
# Using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
# hue
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

Violinplot = (Boxplot + KDEplot)

```python
# violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
# hue

sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

```python
# BARPLOT
# some issue with errorbar
import numpy as np
sns.barplot(data=tips, x='sex', y='total_bill',hue='smoker',estimator=np.min,errorbar='sd')
```

```python
sns.barplot(data=tips, x='sex', y='total_bill',ci=None)
```

```python
# point plot
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker',ci=None)
```

```python
# countplot
sns.countplot(data=tips,x='sex',hue='day')
```

```python
# faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='box',row='time')
```
